```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Patient_ID": [201,202,203,204,205,206,207,208,209,210],
    "Name": ["Raj","Neha","Amit","Priti","Kunal","Riya","Arjun","Meena","Suresh","Kavya"],
    "Age": [34,45,None,50,29,41,55,37,48,None],
    "Height_cm": [172,160,175,None,168,162,170,158,174,165],
    "Weight_kg": [70,62,78,68,None,65,75,59,82,63],
    "Diagnosis": ["BP","Diabetes","Diabetes","Heart","BP","BP","Heart","BP","Diabetes","BP"],
    "Sugar_Level": [108,185,None,150,98,115,160,105,170,None]
}

df = pd.DataFrame(data)
df
```

```python
df
```

```python
df.mean(numeric_only=True)
```

```python
plt.scatter(df["Age"], df["Sugar_Level"])
plt.xlabel("Age")
plt.ylabel("Sugar Level")
plt.title("Age vs Sugar Level")
plt.show()
```

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Age"]
```

```python
avg_sugar = df.groupby("Diagnosis")["Sugar_Level"].mean()

avg_sugar.plot(kind="bar")
plt.xlabel("Diagnosis")
plt.ylabel("Average Sugar Level")
plt.title("Average Sugar Level by Diagnosis")
plt.show()
```

```python


plt.figure()
plt.boxplot(df["Sugar_Level"].dropna())
plt.ylabel("Sugar Level")
plt.title("Box Plot of Sugar Level")
plt.show()
```

```python
plt.hist(df["Age"].dropna())
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```
